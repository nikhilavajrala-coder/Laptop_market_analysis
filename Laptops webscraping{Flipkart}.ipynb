```python
Brand = []
Processor = []
RAM = []
SSD = []
Display = []
Price = []
Discount = []
Rating = []
```

```python
for k in range(1,4):
    url = "https://www.flipkart.com/search?q=laptops&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=on&as=off-&pages={k}"

    headers={"User-Agent": "Mozilla/5.0"}
    response = requests.get(url,headers = headers)
    soup = BeautifulSoup(response.text, "html.parser")

    Laptops = soup.select("div.nZIRY7")
    
    for i in Laptops:
        has_ssd = False
        Brand.append(i.select_one("div.RG5Slk").text)
        for j in i.select(".DTBslk"):
            if "Processor" in j.text:
                Processor.append(j.text)
            if "RAM" in j.text:
                RAM.append(j.text)
            if "SSD" in j.text or "ssd" in j.text:
                SSD.append(j.text)
                has_ssd = True
            if "Display" in j.text:
                Display.append(j.text)
        if not has_ssd:   
            SSD.append("NA")
        Price.append(i.select_one("div.hZ3P6w.DeU9vF").text)
        rating_tag = i.select_one(".MKiFS6")
        if rating_tag:
            Rating.append(rating_tag.text)
        else:
            Rating.append("NA")
        discount_tag = i.select_one(".HQe8jr")
        if discount_tag:
            Discount.append(discount_tag.text)
        else:
            Discount.append("NA")
```

```python
print(len(Brand))
print(len(Processor))
print(len(RAM))
print(len(SSD))
print(len(Display))
print(len(Price))
print(len(Rating))
print(len(Discount))
```
